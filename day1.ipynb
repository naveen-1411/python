Operators -
It is used to perform operations on variables and vale

Arithmetic Operators - Arithmetic operators are used with numeric values to perform common mathematical operations:

Operator	 Name	        Example	
  +	       Addition	         x + y	
  -      Subtraction	     x - y	
  *	     Multiplication      x * y	
  /       Division	         x / y	
  %       Modulus	         x % y	
 **	    Exponentiation	    x ** y	
 //	    Floor division	    x // y

```python
x=44
y=66
print("addition ofx and y - ",x+y)
print("subtraction of x and y - ",x-y)
print("Multiplication of x and y - ",x*y)
print("Division of x and y - ",x/y)
print("Modulus (reminder) of x and y - ", x%y)
print("exponentiation of x and y -  ",x**y)
print("floor division of x and y - ",x//y)
```
